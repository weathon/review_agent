```python
import json
with open("iclr2025_titles_abstracts.json", "r") as f:
    data = json.load(f)
```

```python
data = [i for i in data if i["avg_score"] is not None]
```

```python
data = sorted(data, key=lambda x: x["avg_score"])
```

```python
import time
from rank_bm25 import BM25Okapi
import os
print(f"Indexing ...")
path = "../../iclr2025"
assert os.path.exists(path), f"Path {path} does not exist."
start = time.time()
database = {}

all_files = [i["title"] + "\n" + i["abstract"] for i in data]

tokenized_corpus = [doc.split(" ") for doc in all_files if doc.strip()]

bm25 = BM25Okapi(tokenized_corpus)

print("Indexing complete. Time taken: {:.2f}s".format(time.time() - start))

```
